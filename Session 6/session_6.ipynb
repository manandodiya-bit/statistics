## Task 1: Hypotheses Formulation (Instagram vs. YouTube Daily Time)

**Task:** Write down the null hypothesis ($H_0$) and alternative hypothesis ($H_1$) for the following scenario: *You believe that people spend more time on Instagram than on YouTube daily.*

### Parameter Definitions:
- Let $\mu_{\text{Insta}}$ = Population mean daily time spent on Instagram (minutes)
- Let $\mu_{\text{YT}}$ = Population mean daily time spent on YouTube (minutes)

### Hypotheses (One-Tailed / Directional Test):
- **Null Hypothesis ($H_0$):**
  $$H_0: \mu_{\text{Insta}} \le \mu_{\text{YT}} \quad (\text{or } \mu_{\text{Insta}} - \mu_{\text{YT}} \le 0)$$
  *In Plain English:* People spend equal or less daily time on Instagram compared to YouTube (there is no positive difference in daily screen time).

- **Alternative Hypothesis ($H_1$):**
  $$H_1: \mu_{\text{Insta}} > \mu_{\text{YT}} \quad (\text{or } \mu_{\text{Insta}} - \mu_{\text{YT}} > 0)$$
  *In Plain English:* People spend significantly more daily time on Instagram than on YouTube (matches the researcher's claim).

## Task 2: Two-Sample t-Test on Zomato Home Page Redesign

**Task:** Given the data: Group A (users shown a new Zomato home page) had an average order value of ₹350 from 30 users, and Group B (old home page) had an average order value of ₹320 from 30 users. Assume both groups have a standard deviation of ₹50. Calculate the p-value using a two-sample t-test (you may use Python, R, or an online calculator), and decide whether to accept or reject the null hypothesis at a 0.05 significance level.

### Step-by-Step Mathematical Formulation:
- **Group A (New):** $\bar{x}_1 = 350,\; s_1 = 50,\; n_1 = 30$
- **Group B (Old):** $\bar{x}_2 = 320,\; s_2 = 50,\; n_2 = 30$
- **Degrees of Freedom ($df$):** $n_1 + n_2 - 2 = 30 + 30 - 2 = 58$
- **Pooled Standard Deviation ($s_p$):**
  $$s_p = \sqrt{\frac{(30-1)50^2 + (30-1)50^2}{58}} = 50.00$$
- **Standard Error ($SE$):**
  $$SE = s_p \sqrt{\frac{1}{n_1} + \frac{1}{n_2}} = 50 \sqrt{\frac{2}{30}} \approx 12.9099$$
- **t-statistic ($t$):**
  $$t = \frac{\bar{x}_1 - \bar{x}_2}{SE} = \frac{350 - 320}{12.9099} \approx \mathbf{2.3238}$$
- **p-value:** Two-tailed $p \approx \mathbf{0.0237}$ (One-tailed $p \approx \mathbf{0.0118}$)

### Decision at $\alpha = 0.05$:
Since $p\text{-value } (0.0237) < 0.05$, we **REJECT the null hypothesis ($H_0$)**.
**Conclusion:** The new home page design generates a statistically significant increase in average order value.

```python
from scipy import stats

# Two-sample t-test from summary statistics
t_stat, p_val = stats.ttest_ind_from_stats(
    mean1=350, std1=50, nobs1=30,
    mean2=320, std2=50, nobs2=30,
    equal_var=True
)

print(f't-statistic: {t_stat:.4f}')
print(f'p-value (Two-Tailed): {p_val:.4f}')
print(f'p-value (One-Tailed): {p_val / 2:.4f}')

if p_val < 0.05:
    print('Decision: REJECT H0 (p < 0.05). Significant increase in order value!')
else:
    print('Decision: FAIL TO REJECT H0 (p >= 0.05).')
```

## Task 4: Flipkart Wishlist Button A/B Test (520 vs 480 Clicks out of 1000)

**Task:** Suppose you run an A/B test for a new 'Add to Wishlist' button design on Flipkart. Out of 1000 users, 520 clicked the new button versus 480 who clicked the old button. Without doing detailed calculations, explain if this difference is likely to be significant or just due to chance. Briefly explain how the p-value helps you decide.

### 1. Is this difference likely to be significant or due to chance?
- **Answer:** It is **LIKELY DUE TO CHANCE (Not statistically significant)**.
- **Reasoning:**
  - Click rates are $52.0\%$ vs $48.0\%$ ($4\%$ percentage point difference).
  - In a sample of $1,000$ users, a $52\% - 48\%$ split is typical random sampling noise (identical to flipping a fair coin $1,000$ times and getting $520$ heads).
  - The standard error for proportion with $n = 1000$ is $\approx 1.58\%$, meaning a $2\%$ deviation from $50\%$ is well within normal $\pm 2$ standard error natural variation ($z \approx 1.79$).

### 2. How the p-value helps you decide:
- The **p-value** measures the exact probability of observing a difference of 40 clicks or larger purely by random luck when both buttons are actually equally effective ($H_0$ is true).
- Here, the two-proportion z-test yields $p \approx \mathbf{0.0736}$ ($7.36\%$), which is **greater than $\alpha = 0.05$**.
- Since $p > 0.05$, there is more than a $7\%$ chance this result is just random noise. We fail to reject $H_0$, preventing Flipkart from wasting engineering effort rolling out a redesign that provides no true measurable advantage.

```python
import math
from scipy import stats

# Two-proportion z-test
n = 1000
x1, x2 = 520, 480
p_pool = (x1 + x2) / (2 * n)
se = math.sqrt(p_pool * (1 - p_pool) * (2 / n))
z = (x1/n - x2/n) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print('Two-Proportion z-test:')
print(f'  z-score: {z:.4f}')
print(f'  p-value: {p_value:.4f}')
print('Decision: Fail to reject H0 (p > 0.05). Difference is NOT statistically significant.')
```
