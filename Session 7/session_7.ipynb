## Task 1: Instagram Hours vs. Posts Created (Scatter Plot)

**Task:** Create two Python lists: one representing the number of hours you spend on Instagram each day for a week, and another for the number of posts you make each day. Plot these as a scatter plot using matplotlib and visually describe whether the relationship looks like a positive, negative, or zero correlation.

### Visual & Statistical Description:
- **Relationship:** **Strong Positive Correlation ($r \approx 0.96$)**.
- **Visual Shape:** The points ascend in a clear upward trajectory from bottom-left to top-right.
- **Reasoning:** As daily time dedicated to Instagram increases from $1.5$ hours to $5.0$ hours, the volume of posts/stories shared increases proportionally from $1$ to $4$ posts.

```python
import matplotlib.pyplot as plt
import numpy as np

instagram_hours = [1.5, 2.0, 3.5, 2.5, 4.0, 5.0, 4.5]
posts_created = [1, 1, 2, 2, 3, 4, 3]

r = np.corrcoef(instagram_hours, posts_created)[0, 1]
print('Instagram Hours (X):', instagram_hours)
print('Posts Made (Y):     ', posts_created)
print(f'Pearson Correlation (r): {r:.4f} (Strong Positive Correlation)')

# Scatter plot
plt.figure(figsize=(7, 4.5))
plt.scatter(instagram_hours, posts_created, color='#E1306C', s=90, edgecolors='black', linewidth=1.2, zorder=3)
m, c = np.polyfit(instagram_hours, posts_created, 1)
x_vals = np.linspace(1, 5.5, 100)
plt.plot(x_vals, m * x_vals + c, color='#833AB4', linestyle='--', linewidth=2, label=f'Trend Line (r = {r:.2f})')
plt.title('Daily Instagram Hours vs. Posts Made')
plt.xlabel('Hours Spent on Instagram')
plt.ylabel('Posts Made')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

## Task 2: Zomato vs. Swiggy Order Count Correlation (`np.corrcoef`)

**Task:** Given the following data for 6 friends: their monthly Zomato order count and their monthly Swiggy order count, calculate the correlation coefficient between the two lists using numpy's corrcoef(). Print the result and explain if the correlation is positive, negative, or zero.

### Data for 6 Friends:
- **Zomato Orders:** `[18, 12, 25, 8, 15, 22]`
- **Swiggy Orders:** `[16, 10, 24, 7, 14, 20]`

```python
zomato_orders = [18, 12, 25, 8, 15, 22]
swiggy_orders = [16, 10, 24, 7, 14, 20]

corr_matrix = np.corrcoef(zomato_orders, swiggy_orders)
r_food = corr_matrix[0, 1]

print('Correlation Matrix:')
print(corr_matrix)
print(f'Correlation Coefficient (r): {r_food:.4f}')
print('Explanation: Strong Positive Correlation (r > 0.99). Friends who order frequently on Zomato also order heavily on Swiggy.')
```

## Task 3: Two Real-Life Variables - Expected Correlation & Why

**Task:** Think of two variables from your daily life (for example, number of YouTube videos watched per day and daily study hours). Write a short paragraph describing whether you expect a positive, negative, or zero correlation between them, and why.

### Variables:
- **Variable $X$:** Daily hours spent on YouTube / non-academic screen entertainment
- **Variable $Y$:** Daily hours spent on focused study and coding practice

### Description & Justification:
> *I expect a **strong negative correlation** ($r < 0$) between daily YouTube entertainment hours and daily study hours. A day provides a finite budget of roughly 16 waking hours. Non-academic video consumption directly consumes both time and cognitive energy that would otherwise be allocated to rigorous study. ## Task 4: Spotify Time vs. Test Marks (Strong Negative Correlation)

**Task:** Use ChatGPT to generate a small dataset (10 pairs of numbers) that shows a strong negative correlation between time spent on Spotify and marks scored in a test. Paste the dataset and plot it as a scatter plot using matplotlib.

### 10 Pairs Dataset:
- **Spotify Hours ($X$):** `[0.5, 1.0, 1.5, 2.0, 3.0, 3.5, 4.0, 4.5, 5.5, 6.0]`
- **Test Marks ($Y$):** `[94, 88, 85, 81, 72, 68, 60, 55, 44, 38]`

```python
spotify_hours = np.array([0.5, 1.0, 1.5, 2.0, 3.0, 3.5, 4.0, 4.5, 5.5, 6.0])
test_marks = np.array([94, 88, 85, 81, 72, 68, 60, 55, 44, 38])

r_neg = np.corrcoef(spotify_hours, test_marks)[0, 1]
print('Spotify Hours:', list(spotify_hours))
print('Test Marks:   ', list(test_marks))
print(f'Calculated Correlation (r): {r_neg:.4f} (Extremely Strong Negative Correlation)')

# Scatter plot with trend line
plt.figure(figsize=(7.5, 4.8))
plt.scatter(spotify_hours, test_marks, color='#1DB954', s=90, edgecolors='black', linewidth=1.2, zorder=3, label='Students')
m, c = np.polyfit(spotify_hours, test_marks, 1)
x_vals = np.linspace(0.2, 6.3, 100)
plt.plot(x_vals, m * x_vals + c, color='#191414', linestyle='-', linewidth=2.2, label=f'Regression Line (r = {r_neg:.4f})')
plt.title('Daily Spotify Hours vs. Test Marks (Strong Negative Correlation)')
plt.xlabel('Daily Spotify Hours')
plt.ylabel('Marks Scored (out of 100)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```
