## Task 1: Vector Representation of Playlist Durations
Create a Python list to represent a playlist of 5 song durations in seconds (e.g., `[210, 180, 245, 200, 175]`). Print the total duration of the playlist.

$$\mathbf{d} = \begin{bmatrix} 210 \\ 180 \\ 245 \\ 200 \\ 175 \end{bmatrix}, \quad T = \sum_{i=1}^{5} d_i$$

```python
# 1D Vector representation
playlist_durations = [210, 180, 245, 200, 175]
durations_vector = np.array(playlist_durations)

total_seconds = int(np.sum(durations_vector))
minutes = total_seconds // 60
seconds = total_seconds % 60

print(f"Playlist Durations: {playlist_durations}")
print(f"Vector Dimension: {durations_vector.ndim}D, Shape: {durations_vector.shape}")
print(f"Total Duration: {total_seconds} seconds ({minutes}m {seconds}s)")
print(f"Mean Track Duration: {np.mean(durations_vector):.1f} seconds")
print(f"L2 Norm (Magnitude): {np.linalg.norm(durations_vector):.2f}")
```

## Task 2: 3x3 Swiggy Food Ratings Matrix
Represent a 3x3 matrix in Python to store the ratings (1-5) given by 3 friends to 3 different food items on Swiggy. Print the matrix and display the rating each friend gave to the second food item.

$$\mathbf{R} = \begin{bmatrix} 5 & 4 & 3 \\ 3 & 5 & 2 \\ 4 & 2 & 5 \end{bmatrix}$$

```python
friends = ["Friend 1 (Rahul)", "Friend 2 (Sneha)", "Friend 3 (Amit)"]
food_items = ["Biryani", "Butter Chicken (Item 2)", "Paneer Tikka"]

# 3x3 Matrix
ratings_matrix = np.array([
    [5, 4, 3],  # Friend 1
    [3, 5, 2],  # Friend 2
    [4, 2, 5]   # Friend 3
])

print("Ratings Matrix:")
print(ratings_matrix)

# Column index 1 represents the second food item
second_item_ratings = ratings_matrix[:, 1]

print(f"\nRatings given to the SECOND food item ('{food_items[1]}'):")
for friend, rating in zip(friends, second_item_ratings):
    print(f"  - {friend}: {rating} / 5")

print(f"\nAverage rating for second food item: {np.mean(second_item_ratings):.2f} / 5")
```

## Task 3: 3x4 Orders Matrix Across Days (Column Sums)
Given a matrix representing the number of orders for 4 different dishes across 3 days (rows: days, columns: dishes), write code to calculate and print the total orders for each dish.

$$\mathbf{O} = \begin{bmatrix} 12 & 18 & 25 & 10 \\ 15 & 22 & 30 & 14 \\ 20 & 25 & 35 & 18 \end{bmatrix}, \quad \mathbf{t}_{\text{dish}} = \sum_{\text{days}} \mathbf{O}_{i, *}$$

```python
days = ["Day 1 (Friday)", "Day 2 (Saturday)", "Day 3 (Sunday)"]
dishes = ["Butter Chicken", "Paneer Tikka", "Masala Dosa", "Hakka Noodles"]

# 3 days x 4 dishes matrix
orders_matrix = np.array([
    [12, 18, 25, 10],
    [15, 22, 30, 14],
    [20, 25, 35, 18]
])

# Column-wise sum across rows (axis=0)
total_orders_per_dish = np.sum(orders_matrix, axis=0)

print("Orders Matrix:")
print(orders_matrix)
print("\nTotal Orders Per Dish:")
for dish, total in zip(dishes, total_orders_per_dish):
    print(f"  - {dish}: {total} orders")

most_ordered_idx = np.argmax(total_orders_per_dish)
print(f"\nMost Popular Dish: {dishes[most_ordered_idx]} ({total_orders_per_dish[most_ordered_idx]} orders)")
```

## Task 4: Flipkart Products Revenue (Price x Quantity Matrix)
Imagine you have a table of 5 Flipkart products with columns for price and quantity sold. Represent this data as a matrix in Python, then write a function to calculate the total revenue for each product.

$$\text{Revenue}_i = \text{Price}_i \times \text{Quantity}_i$$

```python
def calculate_product_revenue(product_matrix: np.ndarray) -> np.ndarray:
    """
    Calculates the total revenue for each product row.
    """
    prices = product_matrix[:, 0]
    quantities = product_matrix[:, 1]
    return prices * quantities

product_names = [
    "Wireless Mouse",
    "Bluetooth Headphones",
    "USB-C Fast Cable",
    "10000mAh Power Bank",
    "Ergonomic Laptop Stand"
]

# 5 products x 2 columns [Price (INR), Quantity Sold]
flipkart_matrix = np.array([
    [499.0, 120.0],
    [1499.0, 85.0],
    [199.0, 310.0],
    [999.0, 65.0],
    [799.0, 40.0]
])

revenues = calculate_product_revenue(flipkart_matrix)
total_catalog_revenue = np.sum(revenues)

print(f"{'Product':<25} | {'Price':<10} | {'Qty':<6} | {'Revenue (INR)':<15}")
print("-" * 62)
for i, name in enumerate(product_names):
    p = flipkart_matrix[i, 0]
    q = int(flipkart_matrix[i, 1])
    rev = revenues[i]
    print(f"{name:<25} | {p:<10.2f} | {q:<6} | INR {rev:<12.2f}")
print("-" * 62)
print(f"{'TOTAL CATALOG REVENUE':<45} | INR {total_catalog_revenue:<12.2f}")
```
