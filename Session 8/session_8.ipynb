## Task 1: Basic Probability Function & Coin Toss

**Task:** Write a Python function `calculate_probability(event_count, total_outcomes)` that returns the probability of an event, and use it to find the probability of getting 'Heads' when tossing a coin.

### Formula:
$$P(E) = \frac{\text{Number of favorable outcomes } (n(E))}{\text{Total number of possible outcomes } (n(S))}$$

For a fair coin:
$$S = \{\text{'Heads'}, \text{'Tails'}\}, \quad n(S) = 2$$
$$E = \{\text{'Heads'}\}, \quad n(E) = 1$$
$$P(\text{'Heads'}) = \frac{1}{2} = 0.50\text{ (50%)}$$

```python
def calculate_probability(event_count, total_outcomes):
    """Calculates probability P(E) = event_count / total_outcomes."""
    if total_outcomes <= 0:
        raise ValueError('Total outcomes must be greater than zero.')
    return event_count / total_outcomes

# Coin toss example
p_heads = calculate_probability(event_count=1, total_outcomes=2)
print(f"Probability of getting 'Heads': {p_heads:.4f} ({p_heads * 100:.1f}%)")
```

## Task 2: Dice Rolling Simulation (100 Rolls) & Experimental Probability

**Task:** Simulate rolling a six-sided dice 100 times using Python's random module, count how many times a '6' appears, and calculate the experimental probability of getting a '6'.

### Comparison:
- **Theoretical Probability:** $P(6) = \frac{1}{6} \approx 0.1667$ ($16.67\%$)
- **Experimental Probability:** $\frac{\text{Frequency of 6s}}{100}$

```python
import random
from collections import Counter

random.seed(42)
dice_rolls = [random.randint(1, 6) for _ in range(100)]
count_6 = dice_rolls.count(6)
exp_prob = count_6 / 100
theo_prob = 1 / 6

print('Rolls breakdown:', dict(sorted(Counter(dice_rolls).items())))
print(f"Count of '6' appeared: {count_6} times")
print(f'Experimental Probability: {exp_prob:.4f} ({exp_prob * 100:.1f}%)')
print(f'Theoretical Probability:  {theo_prob:.4f} ({theo_prob * 100:.1f}%)')
```

## Task 4: Zomato Food & Dessert Orders (Intersection over Union)

**Task:** Write a function that, given two arrays — one for users who ordered food on Zomato and one for users who ordered dessert — calculates the probability that a randomly selected user ordered both food and dessert (i.e., intersection over total unique users).

### Formula:
$$P(\text{Food } \cap \text{ Dessert}) = \frac{|\text{Food} \cap \text{Dessert}|}{|\text{Food} \cup \text{Dessert}|}$$

```python
def probability_both_ordered(food_users, dessert_users):
    set_food = set(food_users)
    set_dessert = set(dessert_users)
    both = set_food.intersection(set_dessert)
    all_users = set_food.union(set_dessert)
    return len(both) / len(all_users) if all_users else 0.0

food = ['user1', 'user2', 'user3', 'user4', 'user5', 'user6', 'user7', 'user8']
dessert = ['user3', 'user4', 'user5', 'user9', 'user10']

prob_both = probability_both_ordered(food, dessert)
print(f'Users Ordered Food:    {len(set(food))} users')
print(f'Users Ordered Dessert: {len(set(dessert))} users')
print(f'Users Ordered Both:    {len(set(food).intersection(set(dessert)))} users')
print(f'Total Unique Users:    {len(set(food).union(set(dessert)))} users')
print(f'Probability: {prob_both:.4f} ({prob_both * 100:.2f}%)')
```

## Task 5: Conditional Probability for Flipkart Recommendations

**Task:** Explain with a short Python code example how conditional probability could be used to improve recommendations in an app like Flipkart (e.g., probability of buying headphones given user bought a phone).

### Mathematical Principle:
$$P(\text{Headphones} \mid \text{Phone}) = \frac{P(\text{Headphones} \cap \text{Phone})}{P(\text{Phone})} = \frac{\text{Count}(\text{Bought Both Phone and Headphones})}{\text{Count}(\text{Bought Phone})}$$

### Practical Application:
If $P(\text{Headphones} \mid \text{Phone}) = 70\%$, Flipkart configures a dynamic recommendation widget:
Whenever a shopper adds a smartphone to their cart, Flipkart instantly displays a high-converting **"Frequently Bought Together" bundle** featuring matching wireless headphones, maximizing average order value.

```python
total_users = 1000
phone_buyers = 300
both_buyers = 210

p_headphones_given_phone = both_buyers / phone_buyers

print('Flipkart Customer Base:', total_users)
print('Phone Buyers:          ', phone_buyers)
print('Bought Both:           ', both_buyers)
print(f'P(Headphones | Phone) = {p_headphones_given_phone:.4f} ({p_headphones_given_phone * 100:.1f}%)')
```
