## Task 2: Music App Sample Data Classification

**Task:** Given this sample data from a music app: `['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']`, classify the type of data and explain your reasoning in 2-3 lines.

### Classification
**Categorical Data (Qualitative - Nominal)**

### Reasoning (2-3 lines)
1. **Qualitative Labels:** The data represents music genre names/labels rather than measurable numeric amounts.
2. **No Inherent Hierarchy:** There is no natural mathematical order or ranking among the categories (e.g., 'Pop' is not mathematically greater than or less than 'Rock').
3. **Non-Arithmetic:** You cannot perform mathematical operations like addition or subtraction on them, though frequency counts and the Mode ('Pop') can be computed.

```python
from collections import Counter
import statistics

music_data = ['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']
counts = Counter(music_data)
mode_genre = statistics.mode(music_data)

print('Data:', music_data)
print('Frequency counts:', dict(counts))
print(f"Mode (Most frequent genre): {mode_genre} (occurs {counts[mode_genre]} times)")
```

## Task 3: Improving an App Feature with Statistics

**Task:** Pick any one feature from an app you use (such as the number of likes on an Instagram post, or the rating of a restaurant on Zomato) and describe how statistics could help improve that feature for users.

### Feature: Restaurant Rating & Recommendation System on Zomato

#### The Problem:
A naive arithmetic average of ratings causes unfair distortions:
- A brand new restaurant with **1 review of 5.0 stars** outranks an iconic restaurant with **2,500 reviews averaging 4.8 stars**.
- Simple averages are vulnerable to sudden outlier attacks (1-star review-bombing or fake 5-star bot rings).

#### How Statistics Solves This:
1. **Bayesian Weighted Average (Empirical Bayes):**
   $$\text{Weighted Rating} = \frac{v \cdot R + m \cdot C}{v + m}$$
   where $v$ is number of reviews, $R$ is restaurant average, $m$ is minimum threshold reviews, and $C$ is platform-wide average rating. This prevents small-sample outliers from skewing rankings.
2. **Outlier Filtering (Trimmed Mean / IQR):**
   Identifies anomalous spikes in negative reviews using z-scores or Interquartile Range (IQR) to detect review-bombing.
3. **Exponential Time-Decay Weighting:**
   Statistically weights recent reviews (past 30-60 days) higher than 3-year-old reviews to reflect current chef, menu, and service standards.

```python
# Simulation of Simple Mean vs Bayesian Weighted Average
platform_avg = 3.8
m_threshold = 50

restaurants = [
    {"name": "Restaurant A (Brand New)", "reviews": 1, "rating": 5.0},
    {"name": "Restaurant B (Popular Spot)", "reviews": 1200, "rating": 4.8}
]

for r in restaurants:
    v = r["reviews"]
    R = r["rating"]
    bayesian = (v * R + m_threshold * platform_avg) / (v + m_threshold)
    print(f"{r['name']}:")
    print(f"  - Simple Average:   {R:.2f} stars ({v} reviews)")
    print(f"  - Bayesian Rating:  {bayesian:.2f} stars (Statistically Adjusted)")
```

## Task 4: Spotify Playlist Recommendation - 3 Statistical Terms

**Task:** Suppose you are building a new playlist recommendation feature for a Spotify-like app. List 3 basic statistical terms (such as mean, median, mode, variance, etc.) that would be useful in analyzing user listening data, and explain each with a one-line example.

### 1. Mode
- **Definition:** The value that appears most frequently in a dataset.
- **Example:** *Finding the user's most frequently played artist in a week (e.g., Mode = 'Arijit Singh') to anchor their personalized Daily Mix playlist.*

### 2. Mean (Average)
- **Definition:** The sum of all numerical values divided by the number of observations.
- **Example:** *Calculating the user's average song tempo (e.g., Mean = 126 BPM) during evening workout sessions to automatically curate upbeat, high-tempo tracks.*

### 3. Variance (or Standard Deviation)
- **Definition:** A measure of the dispersion or spread of data points around their mean.
- **Example:** *Measuring the variance of genres or tempos in a user's library to decide whether to suggest a specialized single-genre mix (low variance) or a diverse Discover Weekly playlist (high variance).*

```python
workout_tempos = [124, 128, 126, 130, 125, 128, 127]
genres_streamed = ['Pop', 'EDM', 'Pop', 'Hip-Hop', 'Pop', 'Rock', 'Pop']

print('Workout Track Tempos (BPM):', workout_tempos)
print(f"Mean Tempo:     {statistics.mean(workout_tempos):.2f} BPM")
print(f"Median Tempo:   {statistics.median(workout_tempos):.2f} BPM")
print(f"Tempo Variance: {statistics.variance(workout_tempos):.2f}")
print(f"Tempo Std Dev:  {statistics.stdev(workout_tempos):.2f}")
print('Streamed Genres:', genres_streamed)
print(f"Mode Genre:     '{statistics.mode(genres_streamed)}'")
```
