## Task 1: Swiggy Daily Order Counts Skewness & Plot

**Task:** Given the daily order counts for a Swiggy delivery partner over 10 days: `[12, 15, 17, 14, 13, 16, 200, 18, 14, 15]`, plot a simple graph (hand-drawn or using any tool) and describe whether the data is left skewed, right skewed, or symmetrical. Explain your reasoning based on the shape.

### Classification:
**Right-Skewed (Positively Skewed)**

### Reasoning:
1. **Peak on Left:** 9 out of 10 days have deliveries tightly clustered between $12$ and $18$ orders, creating a prominent peak at the lower end.
2. **Elongated Right Tail:** The single outlier of $200$ orders stretches the tail extensively to the right.
3. **Relationship:** $\text{Mean } (33.40) > \text{Median } (15.00)$, which is the defining characteristic of positive (right) skewness.

```python
import statistics
import matplotlib.pyplot as plt

orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]
mean_o = statistics.mean(orders)
median_o = statistics.median(orders)

print('Orders:', orders)
print(f'Mean:   {mean_o:.2f} orders')
print(f'Median: {median_o:.2f} orders')
print(f'Mode:   {statistics.multimode(orders)} orders')

# Plotting the distribution
plt.figure(figsize=(8, 4.5))
plt.hist(orders, bins=[10, 20, 50, 100, 150, 210], color='#fc8019', edgecolor='black', alpha=0.8, rwidth=0.85)
plt.axvline(mean_o, color='red', linestyle='--', linewidth=2, label=f'Mean = {mean_o:.2f}')
plt.axvline(median_o, color='blue', linestyle='-', linewidth=2, label=f'Median = {median_o:.2f}')
plt.title('Swiggy Daily Orders Distribution (Right-Skewed)')
plt.xlabel('Orders Delivered')
plt.ylabel('Frequency (Days)')
plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```

## Task 2: Instagram Influencer Followers Mean vs. Median & Skewness

**Task:** Take the following list of Instagram influencer follower counts (in thousands): `[5, 7, 8, 8, 9, 10, 12, 15, 95]`. Calculate the mean and median, then state which is higher and what that tells you about the skewness of the data.

### Calculations:
- $\text{Mean} = \frac{5 + 7 + 8 + 8 + 9 + 10 + 12 + 15 + 95}{9} = \frac{169}{9} \approx \mathbf{18.78\text{k}}$ ($18,778$ followers)
- $\text{Median} = 5^{\text{th}}\text{ item in sorted list} = \mathbf{9.00\text{k}}$ ($9,000$ followers)

### Comparison & Skewness Interpretation:
- **The Mean ($18.78\text{k}$) is significantly higher than the Median ($9.00\text{k}$)**.
- **Skewness:** Because the mean is pulled strongly towards the higher values by the $95\text{k}$ mega-influencer outlier, the dataset is **Right-Skewed (Positively Skewed)**.

```python
followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]
mean_f = statistics.mean(followers)
median_f = statistics.median(followers)

print('Followers (k):', followers)
print(f'Mean:   {mean_f:.2f}k')
print(f'Median: {median_f:.2f}k')
print(f'Comparison: Mean > Median by {mean_f - median_f:.2f}k -> Right-Skewed')
```

## Task 3: Real-World Example of Right-Skewed App Data

**Task:** Find a real-world example of right-skewed data from any app you use (like Zomato order amounts, Paytm wallet balances, or YouTube video views). Briefly describe the dataset and explain why you think it is right skewed.

### Selected Example: **Zomato Customer Order Amounts (Basket Value in INR)**

#### 1. Dataset Description:
A dataset of final bill totals (in INR) across thousands of food orders placed on Zomato across a city on a weekend.

#### 2. Why It Is Right-Skewed:
- **Natural Zero/Minimum Bound:** Orders cannot be negative, and restaurants enforce minimum basket amounts (e.g. INR 120-150).
- **High Frequency Mode (Peak):** Most individual and family meals fall in the range of INR 250 to INR 500.
- **Unbounded Upper Tail:** Rare high-value orders (catering, house parties, luxury restaurants) reach INR 5,000, INR 12,000, or INR 25,000.
- **Result:** These rare high amounts stretch the tail out to the right, causing $\text{Mean} > \text{Median}$.

## Task 4: Flipkart Product Ratings Skewness & Graph

**Task:** Given this dataset of Flipkart product ratings: `[3, 3, 4, 4, 4, 5, 5, 5, 5, 5]`, draw a rough graph and state if the distribution is symmetrical, left skewed, or right skewed. Justify your answer in one line.

### Classification:
**Left-Skewed (Negatively Skewed)**

### One-Line Justification:
> *The distribution is left-skewed because ratings concentrate heavily at the high end (5 stars), stretching a tail towards lower ratings on the left, resulting in $\text{Mean } (4.30) < \text{Median } (4.50) < \text{Mode } (5.00)$.*

```python
from collections import Counter

ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]
mean_r = statistics.mean(ratings)
median_r = statistics.median(ratings)
mode_r = statistics.mode(ratings)
freq = Counter(ratings)

print('Flipkart Ratings:', ratings)
print(f'Mean:   {mean_r:.2f}')
print(f'Median: {median_r:.2f}')
print(f'Mode:   {mode_r:.2f}')
print(f'Relation: Mean ({mean_r:.2f}) < Median ({median_r:.2f}) < Mode ({mode_r:.2f}) -> Left-Skewed')

# Plotting bar chart
plt.figure(figsize=(7, 4))
plt.bar([1, 2, 3, 4, 5], [freq.get(i, 0) for i in range(1, 6)], color='#2874f0', edgecolor='black', alpha=0.8, width=0.55)
plt.axvline(mean_r, color='red', linestyle='--', linewidth=2, label=f'Mean = {mean_r:.2f}')
plt.axvline(median_r, color='green', linestyle='-', linewidth=2, label=f'Median = {median_r:.2f}')
plt.title('Flipkart Ratings Distribution (Left-Skewed)')
plt.xlabel('Star Rating')
plt.ylabel('Count')
plt.legend()
plt.grid(axis='y', linestyle=':', alpha=0.6)
plt.show()
```
