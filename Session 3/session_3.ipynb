## Task 1: Step Counts Variance Hand Calculation

**Task:** Given an array of daily step counts from your fitness app for one week (e.g., `[4500, 7000, 5000, 8000, 4000, 9000, 6000]`), calculate the variance by hand using the variance formula and show each calculation step.

### Step 1: Calculate the Mean ($\bar{x}$)
$$\bar{x} = \frac{\sum x_i}{n} = \frac{4500 + 7000 + 5000 + 8000 + 4000 + 9000 + 6000}{7} = \frac{43500}{7} \approx 6214.29\text{ steps}$$

### Step 2: Calculate Deviations and Squared Deviations

| Day | Steps ($x_i$) | Deviation ($x_i - \bar{x}$) | Squared Deviation $(x_i - \bar{x})^2$ |
|:---:|:---:|:---:|:---:|
| 1 | 4,500 | $-1714.29$ | $2,938,775.51$ |
| 2 | 7,000 | $+785.71$ | $617,346.94$ |
| 3 | 5,000 | $-1214.29$ | $1,474,489.80$ |
| 4 | 8,000 | $+1785.71$ | $3,188,775.51$ |
| 5 | 4,000 | $-2214.29$ | $4,903,061.22$ |
| 6 | 9,000 | $+2785.71$ | $7,760,204.08$ |
| 7 | 6,000 | $-214.29$ | $45,918.37$ |
| **Total** | **43,500** | **0.00** | **$\sum (x_i - \bar{x})^2 = 20,928,571.43$** |

### Step 3: Compute Variance
- **Sample Variance ($s^2$):**
  $$s^2 = \frac{\sum (x_i - \bar{x})^2}{n - 1} = \frac{20928571.43}{6} \approx \mathbf{3,488,095.24}$$
  $$s = \sqrt{3488095.24} \approx \mathbf{1867.64\text{ steps}}$$

- **Population Variance ($\sigma^2$):**
  $$\sigma^2 = \frac{\sum (x_i - \mu)^2}{N} = \frac{20928571.43}{7} \approx \mathbf{2,989,795.92}$$
  $$\sigma = \sqrt{2989795.92} \approx \mathbf{1729.10\text{ steps}}$$

```python
import statistics

steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]
mean_val = statistics.mean(steps)
s_var = statistics.variance(steps)
s_dev = statistics.stdev(steps)

print('Steps:', steps)
print(f'Mean: {mean_val:.2f} steps')
print(f'Sample Variance: {s_var:.2f}')
print(f'Sample Std Dev:  {s_dev:.2f} steps')
```

## Task 2: Python Function `calculate_standard_deviation(scores)`

**Task:** Write a Python function called `calculate_standard_deviation(scores)` that takes a list of exam scores and returns the standard deviation rounded to two decimal places.

```python
import math

def calculate_standard_deviation(scores):
    """
    Calculates the sample standard deviation for a list of exam scores,
    rounded to two decimal places.
    """
    if len(scores) < 2:
        return 0.0
    mean_val = sum(scores) / len(scores)
    sum_sq_diff = sum((x - mean_val) ** 2 for x in scores)
    variance = sum_sq_diff / (len(scores) - 1)
    return round(math.sqrt(variance), 2)

exam_scores = [78, 85, 92, 60, 88, 75, 95]
print('Scores:', exam_scores)
print(f'Calculated Standard Deviation: {calculate_standard_deviation(exam_scores)}')
```

## Task 3: Spending Consistency Comparison on Zomato

**Task:** Imagine two friends track their daily spending on Zomato over 5 days. Friend A spends `[200, 200, 200, 200, 200]` and Friend B spends `[100, 300, 150, 400, 50]`. Calculate the standard deviation for both and explain in 2-3 lines who has more consistent spending and why.

### Calculations:
- **Friend A:** $\text{Mean} = 200$, $\sigma = \mathbf{0.00}$
- **Friend B:** $\text{Mean} = 200$, $s = \sqrt{\frac{(100-200)^2 + (300-200)^2 + (150-200)^2 + (400-200)^2 + (50-200)^2}{4}} = \sqrt{\frac{85000}{4}} = \mathbf{145.77}$

## Task 4: Startup Monthly Salary Spread & Interpretation

**Task:** You are given the monthly salary data of 6 employees in a startup: `[28000, 29000, 27000, 60000, 26500, 27500]`. Calculate the variance and standard deviation, then interpret in 2-3 lines what the high or low spread means for this team.

```python
salaries = [28000, 29000, 27000, 60000, 26500, 27500]
mean_sal = statistics.mean(salaries)
median_sal = statistics.median(salaries)
var_s = statistics.variance(salaries)
sd_s = statistics.stdev(salaries)

print('Salaries:', salaries)
print(f'Mean Salary:   INR {mean_sal:,.2f}')
print(f'Median Salary: INR {median_sal:,.2f}')
print(f'Sample Variance: {var_s:,.2f}')
print(f'Sample Std Dev:  INR {sd_s:,.2f}')
print('\nInterpretation:')
print('1. 5 of the 6 employees earn very similar salaries (INR 26.5k - 29k), but the team has an exceptionally high spread (SD > INR 13,000).')
print('2. This high spread is caused by a single high-earning outlier (INR 60,000, likely a founder or tech lead), inflating the mean.')
print('3. This reveals high internal salary inequality, and shows that the median (INR 27,750) better represents typical team compensation.')
```
