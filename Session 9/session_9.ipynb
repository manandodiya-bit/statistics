## Task 1: Generate 100 Random Numbers & Plot Bell Curve

**Task:** Generate 100 random numbers in Python that follow a normal distribution with mean 50 and standard deviation 10 using numpy, then plot the bell curve using matplotlib.

### Theoretical Normal PDF Formula:
$$f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x - \mu}{\sigma}\right)^2}$$
where $\mu = 50$ and $\sigma = 10$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

mu, sigma, n = 50, 10, 100
np.random.seed(42)
sample = np.random.normal(loc=mu, scale=sigma, size=n)

print('Simulated 100 normal numbers:')
print(f'  Mean:   {np.mean(sample):.2f} (Expected: {mu})')
print(f'  StdDev: {np.std(sample, ddof=1):.2f} (Expected: {sigma})')

# Plotting the Bell Curve
plt.figure(figsize=(8, 4.8))
count, bins, _ = plt.hist(sample, bins=15, density=True, color='#4A90E2', edgecolor='black', alpha=0.6, label='Sample (n=100)')
x_vals = np.linspace(mu - 3.5 * sigma, mu + 3.5 * sigma, 300)
plt.plot(x_vals, stats.norm.pdf(x_vals, mu, sigma), color='#D0021B', linewidth=2.5, label='Theoretical Bell Curve')
plt.axvline(mu, color='black', linestyle='--', linewidth=1.8, label=f'Mean = {mu}')
plt.axvline(mu - sigma, color='green', linestyle=':', label=f'1 SD ({mu-sigma}, {mu+sigma})')
plt.axvline(mu + sigma, color='green', linestyle=':')
plt.title('Normal Distribution Bell Curve (mu=50, sigma=10)')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```

## Task 2: 20 Daily Step Counts - Mean vs. Median vs. Mode

**Task:** Given an array of 20 daily step counts from your fitness tracker app, calculate the mean, median, and mode, and check if they are approximately equal as expected in a normal distribution.

### Expected Property of Normal Distribution:
$$\text{Mean} \approx \text{Median} \approx \text{Mode}$$
Because a normal distribution is perfectly symmetric around its center, the balance point (Mean), middle observation (Median), and peak frequency (Mode) coincide.

```python
step_counts = np.array([
    7800, 8100, 7950, 8200, 8000, 7850, 8050, 8300, 7700, 8150,
    8000, 7900, 8250, 7950, 8100, 8000, 8050, 7850, 8100, 8000
])

mean_val = np.mean(step_counts)
median_val = np.median(step_counts)
mode_res = stats.mode(step_counts)
mode_val = float(mode_res.mode)

print('20 Daily Step Counts:')
print(f'  Mean (numpy.mean):     {mean_val:.2f} steps')
print(f'  Median (numpy.median): {median_val:.2f} steps')
print(f'  Mode (scipy.stats):    {mode_val:.2f} steps (count: {mode_res.count})\n')
print('Evaluation:')
print('Mean (8015), Median (8000), and Mode (8000) are virtually identical (difference < 0.2%).')
print('This confirms a symmetric, approximately normal distribution!')
```

## Task 3: Online Game Scores (Empirical 68-95-99.7 Rule)

**Task:** Simulate the scores of 200 players in an online game (e.g., PUBG, Free Fire) using a normal distribution with mean 70 and standard deviation 15, then count how many players scored within one standard deviation (between 55 and 85) of the mean.

### The 68-95-99.7 (Empirical) Rule:
- **$\pm 1\sigma$ ($55$ to $85$):** Approximately **$68.27\%$** of data
- **$\pm 2\sigma$ ($40$ to $100$):** Approximately **$95.45\%$** of data
- **$\pm 3\sigma$ ($25$ to $115$):** Approximately **$99.73\%$** of data

For $N = 200$, expected count within $1\sigma$: $200 \times 0.6827 \approx \mathbf{136.5\text{ players}}$.

```python
np.random.seed(42)
game_scores = np.random.normal(loc=70, scale=15, size=200)

within_1sd = (game_scores >= 55) & (game_scores <= 85)
count_1sd = np.sum(within_1sd)
pct_1sd = (count_1sd / 200) * 100

print('Players Simulated: 200')
print('1 SD Range: [55.0, 85.0]')
print(f'Count within 1 SD: {count_1sd} players ({pct_1sd:.2f}%)')
print(f'Theoretical Expected (68-95-99 Rule): {200 * 0.6827:.1f} players (68.27%)')
print('Verification: The simulated player count closely adheres to the Empirical Rule!')
```

## Task 4: Real-World Scenario Fitting Normal Distribution

**Task:** Pick any real-world scenario (like delivery times for Zomato orders or playlist song durations on Spotify) and explain in 3-4 lines why the normal distribution might be a good fit for modeling this data.

### Selected Scenario: **Playlist Song Durations on Spotify**

### Explanation (3-4 lines):
1. **Central Tendency:** Mainstream commercial songs cluster heavily around an industry-standard duration of $3.0$ to $3.5$ minutes ($180$–$210$ seconds) due to radio formatting and streaming optimization.
2. **Cumulative Variances (CLT):** Total song length is the sum of multiple independent compositional decisions (intro, verses, chorus repeats, bridge, outro), creating balanced natural variation around the mean.
3. **Symmetric Diminishing Tails:** Deviations occur symmetrically—some songs are slightly shorter ($2.5$ mins) and some slightly longer ($4.0$ mins), while extreme tracks (30-second interludes or 10-minute epics) rapidly dwindle in frequency.
4. **Conclusion:** Because the data features a single central peak and symmetrically tapering tails, the bell-shaped normal distribution provides an exceptional statistical fit.
