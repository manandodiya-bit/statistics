## Task 2: Random Sampling Simulation (`random.sample()`)

**Task:** Write a Python script that simulates random sampling: given a list of 1000 fake user IDs (e.g., user1 to user1000), select 50 random users to simulate a survey sample. Print the selected user IDs.

```python
import random

# Generate population of 1000 fake user IDs
population = [f'user{i}' for i in range(1, 1001)]

# Reproducible random sample of 50 users
random.seed(42)
survey_sample = random.sample(population, 50)

print('Population Size:', len(population), f'({population[0]} to {population[-1]})')
print('Sample Size:    ', len(survey_sample), 'users\n')
print('Selected 50 User IDs:')
print(survey_sample)
```

## Task 3: Swiggy Food Delivery Times - Why Use Sampling?

**Task:** Imagine you are analyzing food delivery times on Swiggy. Explain in 2-3 sentences why you would use sampling instead of collecting data from every single order ever made.

### Explanation:
1. **Computational & Financial Feasibility:** Querying hundreds of millions of historical orders incurs massive cloud database read costs, heavy computational overhead, and excessive latency.
2. **Relevance to Current Logistics:** Historical orders from 3 years ago reflect outdated traffic conditions, older routing algorithms, and defunct restaurants that no longer represent current operations.
3. **Statistical Sufficiency:** Under the Central Limit Theorem, a representative sample of a few thousand recent orders delivers virtually identical statistical precision for average delivery time estimates with minimal resource usage.

## Task 4: Spotify Bollywood Survey - Bias Identification

**Task:** Given this scenario: A survey about music preferences is only sent to users who recently streamed Bollywood songs on Spotify. Identify and explain the type of bias this introduces in the sample.

### Identified Bias:
**Selection Bias** (specifically **Sampling Bias / Undercoverage Bias**)

### Detailed Explanation:
- **Flawed Sampling Frame:** The target population is all general Spotify users, but the sampling frame excludes anyone who did not recently stream Bollywood music.
- **Systematic Exclusion:** Millions of listeners who stream International Pop, Hip-Hop, Punjabi, K-Pop, EDM, or Classical music are excluded from having any probability of selection.
- **Distorted Findings:** The survey results will heavily overestimate overall interest and engagement in Bollywood music, providing inaccurate insights for platform-wide features.

## Task 5: BookMyShow Movie Booking Habits (Two Sampling Types)

**Task:** Choose any two types of sampling (for example: random sampling, stratified sampling, systematic sampling, cluster sampling). For each, describe how you would use it to select a sample of users from BookMyShow to study movie booking habits.

### Method 1: Stratified Sampling
- **Application:** Divide BookMyShow's user base into distinct strata based on **City Tier** (Tier 1 Metros, Tier 2 Cities, Tier 3 Towns) or **Booking Frequency** (Weekly, Monthly, Occasional).
- **Selection Process:** Randomly draw users proportionally from each stratum (e.g., $500$ from Tier 1, $300$ from Tier 2, $200$ from Tier 3).
- **Benefit:** Ensures smaller city habits and ticket price sensitivities are fairly represented without metro cities dominating the dataset.

### Method 2: Systematic Sampling
- **Application:** Sort weekend movie booking transactions chronologically by order timestamp.
- **Selection Process:** Select every $k^{\text{th}}$ booking (e.g., every $100^{\text{th}}$ transaction: $37^{\text{th}}, 137^{\text{th}}, 237^{\text{th}}, \dots$) after choosing a random initial seed between $1$ and $100$.
- **Benefit:** Simple to automate in live streaming pipelines, ensuring uniform sampling across morning matinees, prime-time evening shows, and late-night screenings without time-clustering bias.

```python
# 1. Stratified Demo
city_strata = {
    'Tier 1 Metros': [f'user_T1_{i}' for i in range(1, 101)],
    'Tier 2 Cities': [f'user_T2_{i}' for i in range(1, 101)],
    'Tier 3 Towns':  [f'user_T3_{i}' for i in range(1, 101)]
}
print('Stratified Sampling Demo (3 from each tier):')
for tier, u_list in city_strata.items():
    print(f'  {tier}: {random.sample(u_list, 3)}')

# 2. Systematic Demo
bookings = [f'booking_{i:04d}' for i in range(1, 51)]
k = 10
start = 4
print('\nSystematic Sampling Demo (every 10th booking from index 4):')
print('  Selected Bookings:', bookings[start::k])
```
