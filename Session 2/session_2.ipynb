## Task 1: Mean of Daily Spotify Song Plays (`playlist_stats.py`)

**Task:** Create a Python script called `playlist_stats.py` that takes a list of daily Spotify song play counts (e.g., `[120, 135, 150, 200, 120, 90, 200]`) and calculates the mean (average) number of plays.

### Formula:
$$\text{Mean } (\bar{x}) = \frac{\sum x_i}{n} = \frac{120 + 135 + 150 + 200 + 120 + 90 + 200}{7} = \frac{1015}{7} = 145.00$$

```python
import statistics

daily_plays = [120, 135, 150, 200, 120, 90, 200]

# Manual calculation
mean_manual = sum(daily_plays) / len(daily_plays)
# Built-in statistics module
mean_stat = statistics.mean(daily_plays)

print('Daily Spotify Play Counts:', daily_plays)
print(f'Total Days: {len(daily_plays)}')
print(f'Total Plays: {sum(daily_plays)}')
print(f'Mean (Manual): {mean_manual:.2f} plays/day')
print(f'Mean (statistics.mean): {mean_stat:.2f} plays/day')
```

## Task 2: Median Delivery Time for Zomato Orders

**Task:** Given the following array of delivery times (in minutes) for Zomato orders: `[30, 25, 40, 35, 30, 45, 30]`, write code to find the median delivery time and print it.

### Steps:
1. Sort the array in ascending order: `[25, 30, 30, 30, 35, 40, 45]`
2. Determine sample size $n = 7$ (odd number).
3. Find the middle element at position $\frac{n+1}{2} = 4$ (0-indexed position 3).
4. Median is **30 minutes**.

```python
delivery_times = [30, 25, 40, 35, 30, 45, 30]
sorted_times = sorted(delivery_times)
n = len(sorted_times)
median_val = sorted_times[n // 2]

print('Original Delivery Times:', delivery_times)
print('Sorted Delivery Times:  ', sorted_times)
print(f'Middle Value at index {n // 2}: {median_val} minutes')
print(f'Median Delivery Time:    {statistics.median(delivery_times)} minutes')
```

## Task 3: Most Common Flipkart Rating (Mode Function)

**Task:** Write a function `most_common_rating(ratings)` that takes a list of Flipkart product ratings (e.g., `[5, 4, 4, 3, 5, 4, 2, 4]`) and returns the mode (most frequent rating).

```python
from collections import Counter
import statistics

def most_common_rating(ratings):
    """Takes a list of Flipkart ratings and returns the mode."""
    if not ratings:
        return None
    return statistics.mode(ratings)

ratings_list = [5, 4, 4, 3, 5, 4, 2, 4]
mode_res = most_common_rating(ratings_list)
freq = Counter(ratings_list)

print('Ratings:', ratings_list)
print('Rating Frequencies:', dict(freq.most_common()))
print(f'Most common rating (Mode): {mode_res} stars')
```

## Task 4: YouTube Views Comparison & Outlier Sensitivity

**Task:** Given three lists representing YouTube video views for three different channels, compare the mean, median, and mode for each channel, and decide which channel's data is most affected by outliers.

*Hint: Try using one list with a very high value (e.g., [100, 120, 110, 105, 5000]) to see the effect on mean vs median.*

```python
channels = {
    'Channel A (Consistent)': [1500, 1600, 1550, 1500, 1650, 1500, 1700],
    'Channel B (Moderate)':   [800, 950, 1200, 850, 1100, 800, 900],
    'Channel C (Viral Outlier)': [100, 120, 110, 105, 110, 5000]
}

for name, views in channels.items():
    m = statistics.mean(views)
    med = statistics.median(views)
    mo = statistics.mode(views)
    diff = abs(m - med)
    print(f"{name}: Mean={m:.2f}, Median={med:.2f}, Mode={mo}, |Mean-Median|={diff:.2f}")

print('\nCONCLUSION: Channel C is the most affected by outliers.')
print('A viral video with 5000 views pulls the mean to 924.17, while the median remains steady at 110.00.')
```

## Task 5: Real Dataset from Kaggle (IPL Player Runs) using Pandas

**Task:** Pick any real dataset from Kaggle or UCI (e.g., IPL player runs, Swiggy restaurant ratings, or movie box office collections), load it using pandas, and display the mean, median, and mode for one numeric column of your choice.

```python
import pandas as pd

# Load real IPL Player Runs dataset
df = pd.read_csv('data/ipl_player_runs.csv')
print('IPL Player Runs Dataset (Head):')
print(df[['player_name', 'team', 'matches', 'runs', 'batting_average', 'strike_rate']].head())

# Chosen numeric column: 'runs'
runs_col = df['runs']
mean_val = runs_col.mean()
median_val = runs_col.median()
mode_val = runs_col.mode().tolist()

print(f"\nStatistics for Numeric Column 'runs':")
print(f"  Mean (Average Runs):   {mean_val:.2f}")
print(f"  Median (Middle Value): {median_val:.2f}")
print(f"  Mode:                  {len(mode_val)} distinct values")

print('\nInterpretation:')
print('The mean (449.44) is greater than the median (413.00) due to exceptional high-scoring outliers')
print('like Shubman Gill (890 runs), creating positive skewness.')
```
