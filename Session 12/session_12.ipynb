## Task 1: 2x2 Instagram Matrix Multiplication
Given two 2x2 matrices representing likes and comments on two Instagram posts over two days, write Python code to multiply them and display the resulting matrix.

$$\mathbf{A} = \begin{bmatrix} 120 & 45 \\ 80 & 60 \end{bmatrix}, \quad \mathbf{B} = \begin{bmatrix} 2 & 1 \\ 1 & 3 \end{bmatrix}$$
$$\mathbf{C} = \mathbf{A} \cdot \mathbf{B}$$

```python
# 2x2 Matrices
A = np.array([
    [120, 45],
    [80, 60]
])

B = np.array([
    [2, 1],
    [1, 3]
])

# Matrix multiplication (A @ B)
C = np.matmul(A, B)

print("Matrix A (Instagram Post Engagements):")
print(A)
print("\nMatrix B (Weights):")
print(B)
print("\nResulting Product Matrix C (A @ B):")
print(C)

# Element verification
print(f"\nC[0,0] = (120*2) + (45*1) = {120*2 + 45*1}")
print(f"C[0,1] = (120*1) + (45*3) = {120*1 + 45*3}")
print(f"C[1,0] = (80*2) + (60*1)  = {80*2 + 60*1}")
print(f"C[1,1] = (80*1) + (60*3)  = {80*1 + 60*3}")
```

## Task 2: Linear Regression Prediction Function
Create a simple Python function `linear_regression_predict(x, m, c)` that takes an input value `x`, slope `m`, and intercept `c`, and returns the predicted output `y` for a linear regression equation $y = mx + c$.

```python
def linear_regression_predict(x, m, c):
    """
    Predicts the output y for linear regression equation y = mx + c
    """
    return m * x + c

# Test evaluation
test_y = linear_regression_predict(x=3.0, m=2.5, c=5.0)
print(f"For x=3.0, m=2.5, c=5.0 => y = {test_y}")
```

## Task 4: Slope and Intercept Interpretation in Flipkart Price Prediction

**1. What the Slope ($m$) Represents:**
> In Flipkart's product price model, the **slope ($m$)** represents the expected increase in a product's price for each additional 1-star improvement in customer rating. A positive slope indicates that highly rated products command a premium price because higher customer satisfaction signals superior build quality, durability, and brand reputation. For instance, a slope of INR 150 means that, on average, a 4.5-star product sells for INR 150 more than an otherwise comparable 3.5-star item.

**2. What the Intercept ($c$) Represents:**
> The **intercept ($c$)** represents the estimated baseline price of a product when its rating is zero ($x = 0$). In an e-commerce context, this reflects the underlying fixed cost of manufacturing, packaging, warehousing, and platform logistics that any item incurs regardless of user reviews. While few items on Flipkart have an active rating of 0 stars, the intercept anchors the linear equation at its theoretical starting baseline before rating premiums are applied.
