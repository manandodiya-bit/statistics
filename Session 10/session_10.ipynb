## Task 1: Generic Bayes' Posterior Probability Function

**Task:** Write a Python function `bayes_posterior(prior, likelihood, evidence)` that calculates and returns the posterior probability using Bayes theorem formula for given values.

### Formula:
$$P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}$$
$$\text{Posterior} = \frac{\text{Likelihood} \times \text{Prior}}{\text{Evidence}}$$

```python
def bayes_posterior(prior, likelihood, evidence):
    """
    Calculates the posterior probability P(A|B) using Bayes' Theorem:
    P(A|B) = [P(B|A) * P(A)] / P(B)
    """
    if evidence <= 0:
        raise ValueError('Evidence P(B) must be greater than zero.')
    return (likelihood * prior) / evidence

# Verification example
prior = 0.01
likelihood = 0.90
evidence = (0.90 * 0.01) + (0.05 * 0.99)  # 0.0585
post = bayes_posterior(prior, likelihood, evidence)
print(f'Posterior Probability P(A|B): {post:.4f} ({post * 100:.2f}%)')
```

## Task 2: Swiggy Premium Restaurant Probability Given a Discount

**Task:** Given the following scenario: On Swiggy, 10% of restaurants are 'premium', and 80% of premium restaurants offer a discount, while only 20% of regular restaurants do. If you see a discount, use Bayes theorem in code to find the probability that the restaurant is premium.

### Probability Formulations:
- **Prior:** $P(\text{Premium}) = 0.10$, $P(\text{Regular}) = 0.90$
- **Likelihood:** $P(\text{Discount} \mid \text{Premium}) = 0.80$, $P(\text{Discount} \mid \text{Regular}) = 0.20$
- **Evidence (Law of Total Probability):**
  $$P(\text{Discount}) = (0.80 \times 0.10) + (0.20 \times 0.90) = 0.08 + 0.18 = \mathbf{0.26}$$
- **Posterior:**
  $$P(\text{Premium} \mid \text{Discount}) = \frac{0.80 \times 0.10}{0.26} = \frac{0.08}{0.26} \approx \mathbf{0.3077\text{ (30.77%)}}$$

```python
# Prior Probabilities
p_premium = 0.10
p_regular = 0.90

# Likelihoods
p_disc_given_prem = 0.80
p_disc_given_reg = 0.20

# Total Evidence P(Discount)
p_discount = (p_disc_given_prem * p_premium) + (p_disc_given_reg * p_regular)

# Posterior Probability
p_prem_given_disc = bayes_posterior(p_premium, p_disc_given_prem, p_discount)

print(f'Prior P(Premium): {p_premium:.2f}')
print(f'Evidence P(Discount): {p_discount:.2f}')
print(f'Posterior P(Premium | Discount): {p_prem_given_disc:.4f} ({p_prem_given_disc * 100:.2f}%)')
```

## Task 3: Naive Bayes Email Spam Filter ('free' in Subject Line)

**Task:** Simulate a basic spam detection filter: Given a list of 20 email subjects (some spam, some not), and the probability that 'free' appears in spam vs non-spam emails, write code to calculate the probability an email is spam if it contains 'free', using Bayes theorem.

```python
emails = [
    {'subject': 'Get your free coupons now!', 'is_spam': 1},
    {'subject': 'Meeting rescheduled to 3 PM', 'is_spam': 0},
    {'subject': 'Claim your free trial access today', 'is_spam': 1},
    {'subject': 'Project status update for Q3', 'is_spam': 0},
    {'subject': 'Exclusive free luxury gift card inside', 'is_spam': 1},
    {'subject': 'Invoice for your recent purchase', 'is_spam': 0},
    {'subject': 'Urgent: Verify your account details', 'is_spam': 1},
    {'subject': 'Team lunch this Friday at noon', 'is_spam': 0},
    {'subject': 'Win free cash prizes instantly!', 'is_spam': 1},
    {'subject': 'Doctor appointment confirmation', 'is_spam': 0},
    {'subject': 'Earn extra income from home easily', 'is_spam': 1},
    {'subject': 'Flight ticket itinerary details', 'is_spam': 0},
    {'subject': 'Weekly product development newsletter', 'is_spam': 0},
    {'subject': 'Feel free to check out the shared document', 'is_spam': 0},
    {'subject': 'Your monthly bank statement is ready', 'is_spam': 0},
    {'subject': 'GitHub security alert for repository', 'is_spam': 0},
    {'subject': 'HR policy guidelines update', 'is_spam': 0},
    {'subject': 'Course assignment grades published', 'is_spam': 0},
    {'subject': 'Gym membership renewal reminder', 'is_spam': 0},
    {'subject': 'Coffee chat tomorrow morning?', 'is_spam': 0}
]

n_total = len(emails)
n_spam = sum(1 for e in emails if e['is_spam'] == 1)
n_ham = n_total - n_spam

p_spam = n_spam / n_total
p_ham = n_ham / n_total

spam_free = sum(1 for e in emails if e['is_spam'] == 1 and 'free' in e['subject'].lower())
ham_free = sum(1 for e in emails if e['is_spam'] == 0 and 'free' in e['subject'].lower())

p_free_given_spam = spam_free / n_spam
p_free_given_ham = ham_free / n_ham
p_free = (p_free_given_spam * p_spam) + (p_free_given_ham * p_ham)

p_spam_given_free = (p_free_given_spam * p_spam) / p_free

print(f'Emails: {n_total} ({n_spam} Spam, {n_ham} Ham)')
print(f'P(Spam): {p_spam:.2f}, P(Ham): {p_ham:.2f}')
print(f"P('free' | Spam): {p_free_given_spam:.4f} ({spam_free}/{n_spam})")
print(f"P('free' | Ham):  {p_free_given_ham:.4f} ({ham_free}/{n_ham})")
print(f"Evidence P('free'): {p_free:.4f}")
print(f"Posterior P(Spam | 'free'): {p_spam_given_free:.4f} ({p_spam_given_free * 100:.2f}%)")
```

## Task 4: Real-World Bayes Theorem Example (Instagram Bot Detection)

**Task:** Use ChatGPT or Copilot to generate a real-world Bayes theorem example related to any app you use (Instagram, Zomato, Flipkart, etc.), then write code to solve it and submit both the prompt you used and your code.

### Prompt Submitted:
> *"Generate a real-world Bayes Theorem probability problem for Instagram: calculating the probability that an account is a fake bot given that it leaves a generic repetitive comment like 'Great post!'. Include prior probabilities and likelihoods."*

### Problem Scenario:
- **P(Bot):** $5\%$ of active Instagram accounts are automated spam bots ($P(\text{Bot}) = 0.05$).
- **P(Real):** $95\%$ of accounts are genuine real users ($P(\text{Real}) = 0.95$).
- **P(Generic | Bot):** $70\%$ of bot accounts leave generic comments ($0.70$).
- **P(Generic | Real):** $10\%$ of genuine users leave generic comments ($0.10$).
- **Target:** Compute $P(\text{Bot} \mid \text{Generic Comment})$.

```python
# Instagram Bot Detection using Bayes' Theorem
p_bot = 0.05
p_real = 0.95
p_gen_given_bot = 0.70
p_gen_given_real = 0.10

# Total evidence P(Generic)
p_generic = (p_gen_given_bot * p_bot) + (p_gen_given_real * p_real)

# Posterior P(Bot | Generic)
p_bot_given_gen = (p_gen_given_bot * p_bot) / p_generic

print(f'Evidence P(Generic): {p_generic:.4f}')
print(f'Posterior P(Bot | Generic Comment): {p_bot_given_gen:.4f} ({p_bot_given_gen * 100:.2f}%)')
```
